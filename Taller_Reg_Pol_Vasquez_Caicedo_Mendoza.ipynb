### **Ejercicio 1: Carga y Preparación Inicial de los Datos**

Primero, carguemos las librerías necesarias y el dataset. Nos enfocaremos en un subconjunto de las variables numéricas para mantener el taller manejable, pero el principio se aplica a todo el dataset.

```python
# Importación de librerías estándar
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# Importación de herramientas de scikit-learn
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.preprocessing import PolynomialFeatures, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.metrics import mean_squared_error
```

Mejorar visualización de dataframes y gráficos

```python
# Que muestre todas las columnas
pd.options.display.max_columns = None
# En los dataframes, mostrar los float con dos decimales
pd.options.display.float_format = '{:,.2f}'.format

# Configuraciones para una mejor visualización
sns.set(style='whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)
```

Cargar el dataset

```python
# Usamos el dataset Ames Housing desde su fuente original (requiere internet)
url = 'http://jse.amstat.org/v19n3/decock/AmesHousing.txt'
df = pd.read_csv(url, sep='\t')
df.head()
```

```python
# Para este taller, solo usaremos algunas columnas clave para simplificar.
df_sub = df[['Overall Qual', 'Gr Liv Area', 'SalePrice']].copy()
df_sub.columns = ['OverallQual', 'GrLivArea', 'SalePrice'] # Renombrar para facilidad
df_sub.head()
```

**Explicación de las variables del dataset reducido**

1. SalePrice: Precio de Venta.

Esta es la variable objetivo. Es el precio final por el cual se vendió la propiedad, medido en dólares estadounidenses.

Es una variable ordinal que califica la calidad general del material y el acabado de la casa. Es una de las variables predictoras (X) más importantes.

### **Ejercicio 2: Dividir el conjunto de datos**

El método `.info()` muestra que no hay nulos en nuestro subconjunto. ¡Perfecto! Ahora, definamos nuestras variables `X` (predictoras) e `y` (objetivo) y dividamos los datos.

```python
from sklearn.model_selection import train_test_split

# 1. Seleccionar las variables de interés
X = df_sub[['OverallQual', 'GrLivArea']]
y = df_sub['SalePrice']

# 2. Dividir en conjuntos de entrenamiento y prueba (70% train, 30% test)
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.3,
    random_state=42
)

# 3. Verificar tamaños
print(f"Tamaño del set de Entrenamiento: {X_train.shape[0]} observaciones")
print(f"Tamaño del set de Prueba: {X_test.shape[0]} observaciones")
```

### **Ejercicio 3: Modelo con Sobreajuste**

Vamos a crear un modelo muy complejo para ver qué tan mal puede generalizar.

**Crear un Modelo Polinómico de Grado 5**

Usa `Pipeline` para combinar `PolynomialFeatures` (grado 5), `StandardScaler` y `LinearRegression`. Entrénalo con los datos de entrenamiento.

```python
#Pipeline
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import PolynomialFeatures, StandardScaler
from sklearn.linear_model import LinearRegression

pipeline_poly = Pipeline([
    ("poly_features", PolynomialFeatures(degree=5, include_bias=False)),
    ("scaler", StandardScaler()),
    ("model", LinearRegression())
])
```

```python
#Entrenar
from sklearn.metrics import mean_squared_error
import numpy as np

pipeline_poly.fit(X_train, y_train)

y_train_pred_poly = pipeline_poly.predict(X_train)
y_test_pred_poly = pipeline_poly.predict(X_test)
```

```python
#RMSE
rmse_train_poly = np.sqrt(mean_squared_error(y_train, y_train_pred_poly))
rmse_test_poly = np.sqrt(mean_squared_error(y_test, y_test_pred_poly))

print(f"RMSE en Entrenamiento (Poli-5): ${rmse_train_poly:,.2f}")
print(f"RMSE en Prueba (Poli-5): ${rmse_test_poly:,.2f}")
```

**Pregunta:** ¿Qué indica la diferencia que observas entre el error de entrenamiento y el de prueba? ¿Le recomendarías este modelo a la firma inmobiliaria? ¿Por qué?

La diferencia entre el RMSE de entrenamiento (34,152) y el de prueba (41,101) indica que el modelo presenta cierto grado de sobreajuste. Aunque el modelo logra ajustarse bien a los datos de entrenamiento, su desempeño disminuye al evaluar datos no vistos, lo que sugiere que está capturando no solo la relación subyacente sino también el ruido.

Si bien el sobreajuste no es extremo, la brecha entre ambos errores evidencia una varianza moderada. Por esta razón, no se recomendaría este modelo como solución final para la firma inmobiliaria, ya que podría generar predicciones menos fiables en nuevos datos. Sería preferible optar por un modelo más simple o aplicar técnicas de regularización para mejorar la capacidad de generalización.

### **Ejercicio 4: Aplicar Regularización**

Ahora, vamos a "curar" el sobreajuste. Usaremos los mismos `PolynomialFeatures` de grado 5, pero cambiaremos el modelo de regresión.

**Implementar Regresión Ridge**

Copia el pipeline anterior, pero reemplaza `LinearRegression` con `Ridge(alpha=10)`. `alpha` es la fuerza de la penalización.

```python
#pipeline
from sklearn.linear_model import Ridge

pipeline_ridge = Pipeline([
    ("poly_features", PolynomialFeatures(degree=5, include_bias=False)),
    ("scaler", StandardScaler()),
    ("model", Ridge(alpha=10))
])
```

```python
#Entrenar
pipeline_ridge.fit(X_train, y_train)

y_train_pred_ridge = pipeline_ridge.predict(X_train)
y_test_pred_ridge = pipeline_ridge.predict(X_test)

```

```python
# RMSE
rmse_train_ridge = np.sqrt(mean_squared_error(y_train, y_train_pred_ridge))
rmse_test_ridge = np.sqrt(mean_squared_error(y_test, y_test_pred_ridge))

print(f"RMSE Entrenamiento (Ridge): ${rmse_train_ridge:,.2f}")
print(f"RMSE Prueba (Poli-5 sin regularización): ${rmse_test_poly:,.2f}")
print(f"RMSE Prueba (Ridge): ${rmse_test_ridge:,.2f}")
```

**Interpreta los resultados:**

**INTERPRETACIÓN DE RESULTADOS:** Los resultados muestran una mejora significativa al aplicar regresión Ridge. Mientras que el modelo polinómico sin regularización presentaba un claro sobreajuste (con un RMSE de prueba considerablemente mayor que el de entrenamiento), el modelo Ridge logra reducir esta brecha casi por completo.

En particular, el RMSE de entrenamiento y prueba son prácticamente iguales (alrededor de 35,200), lo que indica una excelente capacidad de generalización. Aunque el error de entrenamiento aumentó ligeramente, el error en el conjunto de prueba disminuyó de forma importante, evidenciando que el modelo ahora captura mejor la relación subyacente sin ajustarse al ruido.

Por lo tanto, este modelo es altamente recomendable para la firma inmobiliaria, ya que ofrece predicciones más fiables y estables al enfrentarse a nuevos datos.

**Implementar Regresión Lasso**

Haz lo mismo, pero ahora con `Lasso(alpha=500, max_iter=10000)`. Lasso necesita un `alpha` más grande (porque la penalización L1 es diferente) y a veces más `max_iter` para converger.

```python
from sklearn.linear_model import Lasso

pipeline_lasso = Pipeline([
    ("poly_features", PolynomialFeatures(degree=5, include_bias=False)),
    ("scaler", StandardScaler()),
    ("model", Lasso(alpha=500, max_iter=10000))
])
```

```python
# Entrenar
pipeline_lasso.fit(X_train, y_train)

# Predicciones
y_train_pred_lasso = pipeline_lasso.predict(X_train)
y_test_pred_lasso = pipeline_lasso.predict(X_test)
```

```python
# RMSE
rmse_train_lasso = np.sqrt(mean_squared_error(y_train, y_train_pred_lasso))
rmse_test_lasso = np.sqrt(mean_squared_error(y_test, y_test_pred_lasso))

print(f"RMSE Entrenamiento (Lasso): ${rmse_train_lasso:,.2f}")
print(f"RMSE Prueba (Ridge): ${rmse_test_ridge:,.2f} <-- mejor hasta ahora")
print(f"RMSE Prueba (Lasso): ${rmse_test_lasso:,.2f}")
```

**Interpreta los resultados:**

**INTERPRETACIÓN DE RESULTADOS:** Los resultados de Lasso muestran un desempeño inferior en comparación con Ridge. Aunque Lasso introduce regularización y simplifica el modelo al eliminar variables menos relevantes, el valor de α=500 parece ser demasiado alto para este caso.

Esto provoca que el modelo pierda información importante, generando un mayor error en el conjunto de prueba. En consecuencia, se evidencia un problema de subajuste, donde el modelo es demasiado simple para capturar adecuadamente la relación entre las variables.

Por esta razón, Ridge sigue siendo la mejor alternativa, ya que logra un mejor equilibrio entre ajuste y generalización, manteniendo un bajo error tanto en entrenamiento como en prueba.

**Selección de Variables con Lasso**

Una de las grandes ventajas de Lasso es que puede eliminar variables. Vamos a ver cuántas de las características polinómicas que creamos (ej. `OverallQual^2`, `GrLivArea^5`, `OverallQual * GrLivArea^4`, etc.) fueron eliminadas.

```python
# Estraer coeficientes
coefs_lasso = pipeline_lasso.named_steps['model'].coef_

```

```python
# Extraer los nombres de las características generadas
# Guíate por el siguiente código: feature_names = pipeline_lasso.named_steps['poly_features'].get_feature_names_out(X.columns)
feature_names = pipeline_lasso.named_steps['poly_features'].get_feature_names_out(X.columns)
```

```python
# Contar cuántos coeficientes son exactamente cero
num_total = len(coefs_lasso)
num_zero = np.sum(coefs_lasso == 0)
num_nonzero = num_total - num_zero
```

```python
# Realizar un print de:
# Número total de características polinómicas generadas
# Número de características eliminadas por Lasso (coef = 0)
# Número de características CONSERVADAS
# Porcentaje de variables eliminadas
# Lista con los nombres de las características conservadas por Lasso (coef != 0)

percent_zero = (num_zero / num_total) * 100

selected_features = [
    feature_names[i] for i in range(num_total) if coefs_lasso[i] != 0
]

print(f"Número total de características polinómicas: {num_total}")
print(f"Número de características eliminadas por Lasso (coef = 0): {num_zero}")
print(f"Número de características CONSERVADAS: {num_nonzero}")
print(f"Porcentaje de variables eliminadas: {percent_zero:.2f}%")

print("\nCaracterísticas CONSERVADAS por Lasso:")
for feat in selected_features:
    print(feat)
```

**Interpreta los resultados:**

**INTERPRETACIÓN DE RESULTADOS:** Los resultados muestran que el modelo Lasso eliminó 20 de las 25 características polinómicas generadas, conservando únicamente 5 variables relevantes. Esto representa una reducción del 80% en la dimensionalidad del modelo.

Este comportamiento evidencia la capacidad de Lasso para realizar selección automática de variables, descartando aquellas que no aportan valor predictivo y simplificando significativamente el modelo. En este caso, el modelo identifica que solo unas pocas combinaciones de las variables originales son suficientes para explicar el precio de la vivienda.

En conjunto, Lasso demuestra ser una herramienta útil para simplificar modelos complejos, especialmente en contextos con muchas variables derivadas como los modelos polinómicos.
